```python
x["incident_date"] = pd.to_datetime(
    x["incident_date"],
    errors="coerce"
)

x["incident_year"] = x["incident_date"].dt.year
x["incident_month"] = x["incident_date"].dt.month
x["incident_day"] = x["incident_date"].dt.day

x = x.drop(columns=["incident_date"])
```

```python
x = pd.get_dummies(x, drop_first=True, dtype=int)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
x = x.dropna(axis=1, how="all")

print("Shape after removing empty columns:", x.shape)
```

```python
x = pd.get_dummies(x, drop_first=True, dtype=int)
```

```python
from sklearn.impute import SimpleImputer
import pandas as pd

imputer = SimpleImputer(strategy="median")

x = pd.DataFrame(
    imputer.fit_transform(x),
    columns=imputer.get_feature_names_out(),
    index=x.index
)

print("Remaining missing values:", x.isnull().sum().sum())
print("Final shape:", x.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

LR = LogisticRegression(max_iter=1000)

LR.fit(x_train, y_train)

print("Model trained successfully!")
```

```python
LR=LogisticRegression(max_iter=1000)
LR.fit(x_train,y_train)
```

```python
#Testing Accuracy
y_pred=LR.predict(x_test)
accuracy=accuracy_score(y_pred,y_test)
print("Testing Accuracy:",accuracy)
```

```python
#training accuracy
y_pred_train=LR.predict(x_train)
accuracy_train=accuracy_score(y_pred_train,y_train)
print("Testing Accuracy:",accuracy_train)
```

```python
#validation Acuuracy
val_accuracy=cross_val_score(LR,
    x_train,y_train,cv=5
)
print("Validation Accuracy: ",val_accuracy.mean())
```

```python
#Random Forest Classifier
RF=RandomForestClassifier(
    n_estimators=200,max_depth=10,min_samples_leaf=10,verbose=1,n_jobs=1
)
```

```python
RF.fit(x_train,y_train)
```

```python
y_pred_rf=RF.predict(x_test)
accuracy_rf=accuracy_score(y_pred_rf,y_test)
print("Testing Accuracy:",accuracy_rf)
```

```python
y_pred_train_rf=RF.predict(x_train)
accuracy_train_rf=accuracy_score(y_pred_train_rf,y_train)
print("Testing Accuracy:",accuracy_train_rf)
```

```python
#Gradient Boosting
gb=GradientBoostingClassifier(
    n_estimators=200,min_samples_leaf=10,learning_rate=0.1,random_state=42
)
gb.fit(x_train,y_train)
```

```python
y_pred_gb=gb.predict(x_test)
accuracy_gb=accuracy_score(y_pred_gb,y_test)
print("Testing Accuracy:",accuracy_gb)
```
